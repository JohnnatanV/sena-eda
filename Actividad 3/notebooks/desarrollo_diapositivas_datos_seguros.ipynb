```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_raw = pd.read_csv('../Anexos/DatosSeguros.csv')
df_raw.head(5)
```

```python
df = df_raw.dropna().drop_duplicates().copy()
```

```python
sns.barplot(data=df, x='region', y='valor_seguro', estimator=sum)
plt.title('Ingreso Total por Región')
plt.xlabel('Región')
plt.ylabel('Total Valor Seguro')
plt.show()
```

```python
sns.histplot(df['edad'], bins=10, kde=True, color='skyblue')
plt.title('Distribución de la Edad de los Asegurados')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.show()
```

```python
sns.boxplot(data=df, x='fumador', y='valor_seguro', palette='Set2', hue='fumador')
plt.title('Distribución del Valor del Seguro por Condición de Fumador')
plt.xlabel('Fumador')
plt.ylabel('Valor Seguro')
plt.show()
```

```python
sns.scatterplot(
    data=df, x='imc', y='valor_seguro', hue='fumador', alpha=0.7
)
plt.title('Relación entre IMC y Valor del Seguro')
plt.xlabel('Índice de Masa Corporal (IMC)')
plt.ylabel('Valor Seguro')
plt.show()
```

```python
correlacion = df[['edad', 'imc', 'hijos', 'valor_seguro']].corr()
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Mapa de Calor - Correlación entre Variables Numéricas')
plt.show()
```
